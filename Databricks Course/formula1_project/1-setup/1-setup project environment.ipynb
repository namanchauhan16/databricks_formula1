### **Setting the environments for Formula-1 Project**
1. Create external location connect to the formula1 container present in Azure storage (creater manually )
2. Create catalog formula1
3. Create schemas inside formula1 catalog
4. Create volume file in the landing schema

1. Access cloud ctorage using the Storage credential and External Location

```sql
%fs
ls 'abfss://formula1@databrickcourseextdl.dfs.core.windows.net/landing'
```

2. Create catalog for formula 1 which will be 

```sql
%sql -- TO VIEW THE PRESENT CATALOGS
SHOW CATALOGS;
```

```sql
%sql -- TO CREATE THE DESIRED CATALOG AND PROVIDE THE MANAGED LOCATION FOR THE CONTAINER

CREATE CATALOG  IF NOT EXISTS  FORMULA1
    MANAGED LOCATION 'abfss://formula1@databrickcourseextdl.dfs.core.windows.net/'
    COMMENT 'THIS IS THE MAIN CATALOG FOR THE FORMULA PROJECT';
```

Creating schemas landing, bronze, silver and gold

```sql
%sql
CREATE SCHEMA IF NOT EXISTS formula1.landing;
CREATE SCHEMA IF NOT EXISTS formula1.bronze
    MANAGED LOCATION 'abfss://formula1@databrickcourseextdl.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS formula1.silver
    MANAGED LOCATION 'abfss://formula1@databrickcourseextdl.dfs.core.windows.net/silver';
CREATE SCHEMA IF NOT EXISTS formula1.gold
    MANAGED LOCATION 'abfss://formula1@databrickcourseextdl.dfs.core.windows.net/gold';
```

Create volume files

```sql
%sql
CREATE EXTERNAL VOLUME formula1.landing.files
LOCATION 'abfss://formula1@databrickcourseextdl.dfs.core.windows.net/landing';
```
