1. Read file using spark dataframe reader api
2. Add metadata columns
    - source file
    - injestion timestamp
3. write to bronze delta table

```python
source_file = f"{landing_folfer_path}/{v_batch_id}/circuits.csv"
table_name = f"{catalog_name}.{bronze_schema}.circuits"
```

1. Read csv file

```python
# creating a schema where defining the data types for the columns present in the csv

from pyspark.sql.types import StructType, StructField, StringType, DoubleType

circuits_schema = StructType([
    StructField("circuitId", StringType()),
    StructField("url", StringType()),
    StructField("circuitName", StringType()),
    StructField("lat", DoubleType()),
    StructField("long", DoubleType()),
    StructField("locality", StringType()),
    StructField("country", StringType())
])
```

```python
circuits_df = (
    spark.read.format('csv')
    .option('header', True)
    .option('mode', 'FAILFAST')
    .schema(circuits_schema)
    .load(source_file)
    )
```

2. Add metadata columns

```python
circuits_df_final = add_ingestion_metadata(circuits_df)
```

3. Write to bronze delta table

```python
circuits_df_final = circuits_df_final.withColumn("batch_id", F.lit(v_batch_id))
```

```python

(
    circuits_df_final.write
    .format('delta')
    .mode('overwrite')
    .partitionBy('batch_id')
    .option('replaceWhere', f"batch_id='{v_batch_id}'")
    .saveAsTable(table_name)
)
```
