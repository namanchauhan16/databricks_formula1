```python
source_file = f"{landing_folfer_path}/{v_batch_id}/races.csv"
table_name = f"{catalog_name}.{bronze_schema}.races"
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DateType

races_schema = StructType([
    StructField("season", IntegerType()),
    StructField("round", IntegerType()),
    StructField("url", StringType()),
    StructField("raceName", StringType()),
    StructField("date", DateType()),
    StructField("circuitId", StringType())
])
```

```python
races_df = (
    spark.read.format('csv')
    .option('header', True)
    .schema(races_schema)
    .load(source_file)
)
```

```python
races_df_final = add_ingestion_metadata(races_df)
```

```python
races_df_final = races_df_final.withColumn("batch_id", F.lit(v_batch_id))
```

```python
(
    races_df_final.write
    .format('delta')
    .mode('overwrite')
    .partitionBy('batch_id')
    .option('replaceWhere', f"batch_id='{v_batch_id}'")
    .saveAsTable(table_name)
)
```
