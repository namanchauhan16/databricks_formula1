```sql
CREATE OR REPLACE VIEW formula1.gold.v_drivers_standings
AS
WITH drivers_session_summery
AS
(
SELECT 
    f.season,
    d.driver_id,
    d.driver_name,
    d.nationality,
    COUNT(*) AS race_starts,
    SUM(F.points) AS total_points,
    COUNT_IF(F.is_win) AS number_of_wins,
    COUNT_IF(F.is_podium) AS number_of_podiums
FROM 
    formula1.gold.fact_session_table f
JOIN 
    formula1.gold.dim_drivers d
    ON f.driver_id = d.driver_id
GROUP BY
    f.season,
    d.driver_id,
    d.driver_name,
    d.nationality
)
SELECT
    RANK() OVER(PARTITION BY season
                ORDER BY total_points DESC,
                 number_of_wins DESC) AS standing,
    *
FROM drivers_session_summery;
```
